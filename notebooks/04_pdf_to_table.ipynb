```python
import pymupdf
import pandas as pd

doc = pymupdf.open("../data/raw/plan1.pdf")
print("pages:", len(doc))
```

```python
rows = [{"page": i, "text": page.get_text()} for i, page in enumerate(doc, start=1)]
pages = pd.DataFrame(rows)
pages.head()
```

```python
pages["n_chars"] = pages["text"].str.len()
print(pages[["page", "n_chars"]])
print("pages with almost no text:", (pages["n_chars"] < 50).sum())
```

```python
print(pages.loc[1, "text"])
```

```python
print(repr(pages.loc[1, "text"][:300]))
```

```python
from collections import Counter

line_counts = Counter()
for text in pages["text"]:
    for line in set(text.split("\n")):
        line_counts[line.strip()] += 1

for line, count in line_counts.most_common(8):
    print(count, "|", line)
```

## What is messy in plan1.pdf

- Header "NorthStar Freight ... CONFIDENTIAL" repeats on every page
- Footer line and "Page N" repeat on every page
- Sentences are broken across lines with \n
- The table on page 4 comes out as one number per line, not as rows
- The cover page (page 1) has very little text
- Numbers and units appear as "Rs. 42.5 crore", "9.2%", "USD 65"
