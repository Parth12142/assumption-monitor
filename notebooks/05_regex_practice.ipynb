```python
import re

s = "In 2025 the company earned revenue of Rs. 42.5 crore and an operating margin of 9.2%."

print(re.findall(r"\d+", s))
print(re.findall(r"\d+(?:\.\d+)?", s))
print(re.findall(r"\d+(?:\.\d+)?\s?%", s))
print(re.search(r"Rs\.\s?\d+(?:\.\d+)?", s).group())
print(re.sub(r"\s+", " ", "too    many\n spaces"))
print(bool(re.match(r"^Page\s+\d+$", "Page 3")))
print(bool(re.match(r"^Page\s+\d+$", "Page three")))
```
