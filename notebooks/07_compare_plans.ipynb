```python
import pandas as pd

a = pd.read_csv("../data/clean/plan1_sentences.csv")
b = pd.read_csv("../data/clean/plan2_sentences.csv")
both = pd.concat([a, b], ignore_index=True)

print(both.groupby("doc_id").size())
print(both.groupby("doc_id")["extraction_status"].value_counts())
```

```python
dups = both[both.duplicated("sentence", keep=False)]
print(dups[["doc_id", "sentence"]])
```
