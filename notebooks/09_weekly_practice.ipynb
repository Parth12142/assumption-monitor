```python
import numpy as np
import pandas as pd

# 1. Weekly resampling: weeks end on Friday
s = pd.Series(range(1, 16), index=pd.bdate_range("2026-01-05", periods=15))
print(s.resample("W-FRI").mean())          # 3.0, 8.0, 13.0

# 2. Median vs mean with one wild value
x = pd.Series([10, 11, 9, 10, 100], index=pd.bdate_range("2026-01-05", periods=5))
print(x.mean(), x.median())                # 28.0  10.0

# 3. MAD: the median of the distances from the median
mad = (x - x.median()).abs().median()
print(mad)                                 # 1.0
print((x - x.median()) / (1.4826 * mad))   # last value: about 60.7
```
